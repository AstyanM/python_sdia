 Define a function `triangle_shape(height)`, that returns

 - `""` if `height=0`,

 - otherwise it returns a string that forms a triangle with height prescribed by `height`.



 **Examples**



 height=1

 ```

 x

 ```



 height=2

 ```

  x

 xxx

 ```



 height=3

 ```

   x

  xxx

 xxxxx

 ```



 height=4

 ```

    x

   xxx

  xxxxx

 xxxxxxx

 ```



 height=5

 ```

     x

    xxx

   xxxxx

  xxxxxxx

 xxxxxxxxx

 ```



 height=6

 ```

      x

     xxx

    xxxxx

   xxxxxxx

  xxxxxxxxx

 xxxxxxxxxxx

 ```

````python
def triangle_shape(height):
    if height == 0:
        return ""
    lines = []
    for i in range(height):
        spaces = " " * (height - i - 1)
        xs = "x" * (2 * i + 1)
        lines.append(spaces + xs + spaces)
    return "\n".join(lines)

print(triangle_shape(5))
````
